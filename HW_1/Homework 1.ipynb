### 1. How many men and women (sex feature) are represented in this dataset?

```python
# You answer (code + results)

# Count the occurrences of each unique value in the "sex" column
sex_counts = data[" sex"].value_counts().reset_index()
sex_counts.columns = ["Sex", "Count"]
print(sex_counts)
```

### 2. What is the average age (age feature) of women?

```python
# You answer (code + results)

# Filter the DataFrame to include only rows where "sex" is "female"
female = data[data[" sex"] == " Female"]

# Select the "age" column for the filtered rows
female_ages = female["age"]

# find average
average_age_women = female_ages.mean()

print(f"The average age of women is {average_age_women:}")
```

### 3. What is the percentage of German citizens (native-country feature)?


```python
# You answer (code + results)

# Filter rows where "native-country" is "Germany"
german_count = len(data[data[" native-country"] == " Germany"])

# find percentage of germans within the whole data set
german_percentage = (german_count / len(data)) * 100

print(f"The percentage of German citizens is {german_percentage:.2f}%")
```

###  4. What are the mean and standard deviation of age for those who earn more than 50K per year (salary feature) and those who earn less than 50K per year?

```python
# You answer (code + results)

# mean and standard deviation of age for salary > 50K
mean_above_50 = data[data[" salary"] == " >50K"]["age"].mean()
std_above_50 = data[data[" salary"] == " >50K"]["age"].std()

# mean and standard deviation of age for salary  <=50K
mean_below_50 = data[data[" salary"] == " <=50K"]["age"].mean()
std_below_50 = data[data[" salary"] == " <=50K"]["age"].std()

print(f"Mean age for those who earn more than 50K per year: {mean_above_50:.2f}")
print(f"Standard deviation for those who earn more than 50K per year: {std_above_50:.2f}")
print()
print(f"Mean age for those who earn 50K or less per year: {mean_below_50:.2f}")
print(f"Standard deviationfor those who earn 50K or less per year: {std_below_50:.2f}")



```

### 5. Is it true that people who earn more than 50K have at least high school education? (education – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate feature)

```python
# You answer (code + results)

# those who earn above 50k
above_50_earners = data[data[" salary"] == " >50K"]

# education at least high school
edu_lvl = [" Bachelors", " Prof-school", " Assoc-acdm", " Assoc-voc", " Masters", " Doctorate"]

# the number of earners above 50K have that have at least a HS education
above_50_with_education = above_50_earners[above_50_earners[" education"].isin(edu_lvl)]

pct_above_50_with_education = (len(above_50_with_education) / len(above_50_earners)) * 100 

if pct_above_50_with_education == 100:
    print("True\n")
else:
    print("False\n")

print(f"No, not everyone who earns more than 50k has at least a HS education. The percentage of people who do is: {pct_above_50_with_education:.2f}%")
```

### 6.  Display age statistics for each race (race feature) and each gender (sex feature). 

Hint: Use `groupby()` and `describe()` functions of DataFrame. Find the maximum age of men of Amer-Indian-Eskimo race.

```python
# You answer (code + results)

# age statistics for each race and gender
age_statistics = data.groupby([" race", " sex"])["age"].describe()

Amer_Indian_Eskimo = age_statistics.loc[" Amer-Indian-Eskimo"]
Amer_Indian_Eskimo_Male = Amer_Indian_Eskimo.loc[" Male"]
Amer_Indian_Eskimo_Male_Max = Amer_Indian_Eskimo_Male.loc["max"]

print(age_statistics)
print()
print(f"The maximum age of men in Amer-Indian-Eskimo race is: {Amer_Indian_Eskimo_Male_Max}")
```

### 7. What is the maximum number of hours a person works per week (hours-per-week feature)? How many people work such a number of hours, and what is the percentage of those who earn a lot (>50K) among them?


```python
# You answer (code + results)
# 1.
# the maximum hours worked per week
max_hours = data[" hours-per-week"].max()
print(f"The maximum hours worked per week: {max_hours}\n")

# 2.
# how many people work that maximum number of hours
num_ppl_max_hrs = len(data[data[" hours-per-week"] == max_hours])
print(f"The number of people working the maximum hours per week: {num_ppl_max_hrs}\n")

# 3. % of max hour workers that make 50 or more
max_hour_workers = data[data[" hours-per-week"] == max_hours]
# count of how many people make over 50k
above_50_earners = max_hour_workers[max_hour_workers[" salary"] == " >50K"]
num_high_earners = len(above_50_earners)

# percent of people who work the max hours and make over 50k
percentage_above_50_earners = (num_high_earners / num_ppl_max_hrs) * 100

print(f"The percent of people who make more than 50k and work max hours is: {percentage_above_50_earners:.2f}%")

```
